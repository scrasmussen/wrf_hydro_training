## Configure Using Newer CMake Method

````python
./configure_new -x \
    -p gfortran/gcc \
    -- -DWRF_CORE=ARW -DWRF_NESTING=BASIC \
    -DENABLE_HYDRO=ON -DWRF_CASE=EM_REAL
 ./compile_new -j4 |& tee build_new.txt
````
